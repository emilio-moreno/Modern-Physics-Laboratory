```python
import pandas as pd
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

plt.rcParams.update({"font.family": "Times New Roman", "mathtext.fontset": "cm"})
```

```python
workbook = pd.read_excel("2026-10-09_HallEffect.xlsx", sheet_name=["VvsB1", "VvsB2", "VvsB3"])
workbook["VvsB3"].head(3)
```

```python
fig, ax = plt.subplots()

for i, sheet in enumerate(workbook.keys()):
    # Data points
    print(sheet)
    data = workbook[sheet]
    probe_current = data["I_P[mA]"]
    field_current = data["I_B[A]"]
    hall_voltage = data["V_H[mV]"]
    n, e, t = 1, 1, 1

    hall_voltage_current = hall_voltage / probe_current

    ax.scatter(field_current, hall_voltage_current, label=f"Medición {i + 1}")

    # Curve fit
    fit_function = lambda Ib, m: m * (Ib / (n*e*t))
    popt, pcov = curve_fit(fit_function, field_current, hall_voltage_current)
    Ib = np.linspace(np.min(field_current), np.max(field_current), 100)
    ax.plot(Ib, fit_function(Ib, *popt), label=f"Ajuste lineal {i + 1}")
ax.set(xlabel="Corriente del electroimán ($I_B$) [mA]",
       ylabel="Cociente de voltaje de Hall\ncon corriente de sonda ($V_H / I_P$) [V/A]",
       title="Voltaje de Hall variando corriente del electroimán ($I_B$)")
plt.legend()
```

```python
data = workbook["VvsB3"]
probe_current = data["I_P[mA]"]
field_current = data["I_B[A]"]
hall_voltage = data["V_H[mV]"]
n, e, t = 1, 1, 1

field_probe_current = probe_current * field_current

fig, ax = plt.subplots()
ax.scatter(field_probe_current, hall_voltage)
ax.set(xlabel="Corriente del electroimán $\\times$ corriente de sonda ($I_B I_P$) [mA$^2$]",
       ylabel="Voltaje de Hall ($V_H$) [mV]",
       title="Voltaje de Hall variando corriente del electroimán ($I_B$)")
```
